```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# Set seed for reproducibility
np.random.seed(42)
random.seed(42)

print("=" * 60)
print("FRESHBITE BAKERY - DATA GENERATOR")
print("=" * 60)
```

```python
#1. PRODUCTS TABLE

print("\n[1/3] Generating products data...")

product_ids = [f'P{str(i).zfill(3)}' for i in range(1, 46)]

product_names = [
    'Sourdough Bread', 'Sourdough', 'Whole Wheat Bread', 'Baguette', 
    'Croissant', 'Pain au Chocolat', 'Cinnamon Roll', 'Danish Pastry',
    'Blueberry Muffin', 'Chocolate Chip Muffin', 'Banana Bread',
    'Carrot Cake Slice', 'Chocolate Cake Slice', 'Red Velvet Cupcake',
    'Vanilla Cupcake', 'Apple Pie Slice', 'Cherry Pie Slice',
    'Chocolate Chip Cookie', 'Oatmeal Cookie', 'Sugar Cookie',
    'Brownie', 'Lemon Bar', 'Eclair', 'Macaron Box',
    'Sandwich - Turkey', 'Sandwich - Veggie', 'Sandwich - Ham',
    'Bagel - Plain', 'Bagel - Everything', 'Bagel - Sesame',
    'Pretzel', 'Focaccia', 'Ciabatta Roll', 'Pumpernickel Bread',
    'Rye Bread', 'Brioche Bun', 'Kaiser Roll', 'Dinner Roll',
    'French Bread', 'Multigrain Bread', 'Gluten Free Bread',
    'Spinach Quiche', 'Cheese Danish', 'Almond Croissant', 'Fruit Tart'
]

categories = [
    'Bread', 'Bread', 'Bread', 'Bread',
    'Pastry', 'Pastry', 'Pastry', 'Pastry',
    'Muffin', 'Muffin', 'Bread',
    'Cake', 'Cake', 'Cupcake',
    'Cupcake', 'Pie', 'Pie',
    'Cookie', 'Cookie', 'Cookie',
    'Bar', 'Bar', 'Pastry', 'Cookie',
    'Sandwich', 'Sandwich', 'Sandwich',
    'Bagel', 'Bagel', 'Bagel',
    'Bread', 'Bread', 'Bread', 'Bread',
    'Bread', 'Bread', 'Bread', 'Bread',
    'Bread', 'Bread', 'Bread',
    'Savory', 'Pastry', 'Pastry', 'Pastry'
]

# Shelf life in days (with some missing values)
shelf_life_days = [
    3, 3, 4, 2, 1, 1, 2, 2,
    3, 3, 4,
    2, 2, 2,
    2, 3, 3,
    5, 5, 5,
    4, 3, 1, 7,
    1, 1, 1,
    3, 3, 3,
    2, 3, 2, 4,
    4, 2, 2, 3,
    2, 4, 3,
    1, 1, 1, 2
]

# Add some missing shelf life values
for i in [5, 12, 23, 34]:
    shelf_life_days[i] = None

# Unit cost with some variation
base_costs = np.random.uniform(0.5, 4.5, len(product_ids))
unit_costs = np.round(base_costs, 2)

# Selling price (2-3x markup with some variation)
selling_prices = np.round(unit_costs * np.random.uniform(2.2, 3.5, len(product_ids)), 2)

products_df = pd.DataFrame({
    'product_id': product_ids,
    'product_name': product_names,
    'category': categories,
    'shelf_life_days': shelf_life_days,
    'unit_cost': unit_costs,
    'selling_price': selling_prices
})

print(f"✓ Generated {len(products_df)} products")
```

```python
# 2. SALES TRANSACTIONS TABLE

print("\n[2/3] Generating sales transactions...")

store_locations = ['Downtown', 'Westside', 'Eastside']
start_date = datetime(2024, 1, 1)
end_date = datetime(2024, 9, 30)

sales_data = []
transaction_id = 1000

# Generate sales for 9 months
current_date = start_date
while current_date <= end_date:
    day_of_week = current_date.weekday()
    
    for store in store_locations:
        # Weekend = more sales
        if day_of_week >= 5:
            num_transactions = random.randint(25, 45)
        else:
            num_transactions = random.randint(15, 30)
        
        for _ in range(num_transactions):
            # Pick random product (popular items sell more)
            if random.random() < 0.4:
                product_idx = random.choice([0, 1, 4, 8, 17])  # Popular items
            else:
                product_idx = random.randint(0, len(product_ids)-1)
            
            product_id = product_ids[product_idx]
            quantity = random.choices([1, 2, 3, 4], weights=[0.6, 0.25, 0.1, 0.05])[0]
            
            # Add some missing store locations (data quality issue)
            if random.random() < 0.02:
                store_name = None
            else:
                store_name = store
            
            sales_data.append({
                'transaction_id': transaction_id,
                'sale_date': current_date.strftime('%Y-%m-%d'),
                'store_location': store_name,
                'product_id': product_id,
                'quantity_sold': quantity
            })
            transaction_id += 1
    
    current_date += timedelta(days=1)

sales_df = pd.DataFrame(sales_data)
print(f"✓ Generated {len(sales_df)} sales transactions")
```

```python
# 3. INVENTORY RECORDS TABLE

print("\n[3/3] Generating inventory records...")

inventory_data = []
record_id = 5000

current_date = start_date
while current_date <= end_date:
    for store in store_locations:
        # Each store tracks inventory for random products each day
        products_tracked = random.sample(product_ids, k=random.randint(15, 25))
        
        for product_id in products_tracked:
            stock_received = random.choices([0, 5, 10, 15, 20, 30], 
                                           weights=[0.3, 0.2, 0.2, 0.15, 0.1, 0.05])[0]
            
            stock_start = random.randint(0, 30)
            
            # Spoilage happens more on older items
            if stock_start > 15:
                spoilage = random.randint(1, 8)
            elif stock_start > 5:
                spoilage = random.randint(0, 4)
            else:
                spoilage = random.randint(0, 2)
            
            # Add some outliers (data entry errors)
            if random.random() < 0.01:
                spoilage = random.randint(50, 100)
            
            # Add some negative values (data quality issue)
            if random.random() < 0.01:
                spoilage = random.randint(-5, -1)
            
            # Some missing spoilage data
            if random.random() < 0.05:
                spoilage = None
            
            inventory_data.append({
                'record_id': record_id,
                'record_date': current_date.strftime('%Y-%m-%d'),
                'store_location': store,
                'product_id': product_id,
                'stock_start_of_day': stock_start,
                'stock_received': stock_received,
                'stock_spoiled': spoilage
            })
            record_id += 1
    
    current_date += timedelta(days=1)

inventory_df = pd.DataFrame(inventory_data)
print(f"✓ Generated {len(inventory_df)} inventory records")

```

```python
# SAVE TO CSV FILES

print("\n" + "=" * 60)
print("SAVING DATA FILES...")
print("=" * 60)

products_df.to_csv('products.csv', index=False)
print("✓ Saved: products.csv")

sales_df.to_csv('sales_transactions.csv', index=False)
print("✓ Saved: sales_transactions.csv")

inventory_df.to_csv('inventory_records.csv', index=False)
print("✓ Saved: inventory_records.csv")


```
